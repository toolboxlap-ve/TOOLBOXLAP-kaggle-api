The notebook launches the tested TOOLBOXLAP V6 baseline, prints a temporary OpenAI-compatible ngrok URL, and ends with `EVERYTHING IS WORKING` after its public API test.\n\nWebsite: https://toolboxlap.com  |  YouTube: https://www.youtube.com/@TOOLBOXLAP-u1c  |  GitHub: https://github.com/toolboxlap-ve/TOOLBOXLAP-kaggle-api

```python
import getpass
import os
import subprocess
import sys
from pathlib import Path
from urllib.request import urlretrieve

LAUNCHER_URL = 'https://raw.githubusercontent.com/toolboxlap-ve/TOOLBOXLAP-kaggle-api/main/scripts/toolboxlap_kaggle_hf_ollama_ngrok.py'
LAUNCHER_PATH = Path('toolboxlap_kaggle_hf_ollama_ngrok.py')
DEFAULT_MODEL = 'hf.co/HauhauCS/Qwen3.5-9B-Uncensored-HauhauCS-Aggressive:Q4_K_M'

model = input(f'Model [ENTER = default: {DEFAULT_MODEL}]: ').strip() or DEFAULT_MODEL
ngrok_authtoken = getpass.getpass('ngrok authtoken (used only for this Kaggle session): ').strip()
if not ngrok_authtoken:
    raise ValueError('An ngrok authtoken is required to create a public tunnel.')

print('Downloading the current TOOLBOXLAP launcher...')
urlretrieve(LAUNCHER_URL, LAUNCHER_PATH)
launcher_env = os.environ.copy()
launcher_env['TOOLBOXLAP_MODEL'] = model
launcher_env['NGROK_AUTHTOKEN'] = ngrok_authtoken
subprocess.run([sys.executable, str(LAUNCHER_PATH)], check=True, env=launcher_env)
```
